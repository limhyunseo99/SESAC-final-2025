```python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
from reportlab.lib.colors import HexColor
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
from datetime import datetime

# 폰트 등록
pdfmetrics.registerFont(TTFont("Free-Regular", r"c:\USERS\KMOON\APPDATA\LOCAL\MICROSOFT\WINDOWS\FONTS\FREESENTATION-4REGULAR.TTF"))
pdfmetrics.registerFont(TTFont("Free-Bold", r"c:\USERS\KMOON\APPDATA\LOCAL\MICROSOFT\WINDOWS\FONTS\FREESENTATION-7BOLD.TTF"))


def create_cover_page(path: str, country: str, item: str):
    """PDF 표지 생성"""

    w, h = A4
    today = datetime.now().strftime("%Y-%m-%d")

    c = canvas.Canvas(path, pagesize=A4)

    # 배경 설정
    c.setFillColor(HexColor("#FFFFFF"))
    c.rect(0, 0, w, h, fill=1, stroke=0)

    # 큰 원
    big_r = 540
    big_cx = w + 360
    big_cy = h * 0.55
    c.setFillColor(HexColor("#7DA0CA"))
    c.circle(big_cx, big_cy, big_r, fill=1, stroke=0)

    # 작은 원
    small_r = 320
    small_cx = w * 0.70
    small_cy = -40
    c.setFillColor(HexColor("#052659"))
    c.circle(small_cx, small_cy, small_r, fill=1, stroke=0)

    # 제목
    c.setFillColor(HexColor("#052659"))
    c.setFont("Free-Bold", 40)
    c.drawString(70, h - 200, f"{country} 시장 진출 전략 보고서")

    c.setFont("Free-Bold", 48)
    c.drawString(70, h - 250, "2025")

    c.setFont("Free-Regular", 17)
    c.drawString(70, h - 290, "데이터 기반 해외시장 분석")

    # 구분선
    c.setStrokeColor(HexColor("#021024"))
    c.setLineWidth(1)
    c.line(70, h - 305, 320, h - 305)

    # 설명문
    desc = (
        "본 보고서는 국가정보·진출전략·수출데이터 기반으로 생성되었습니다.\n"
        "AI 기반 분석을 통해 시장성 평가 및 진출 전략을 제공합니다."
    )
    c.setFont("Free-Regular", 13)
    y = h - 330
    for line in desc.split("\n"):
        c.drawString(70, y, line)
        y -= 14

    # 하단 발행정보
    c.setFillColor(HexColor("#FFFFFF"))
    c.setFont("Free-Regular", 10)

    base_y = 40

    c.drawRightString(w - 40, base_y + 20, f"발행일: {today}")
    c.drawRightString(w - 40, base_y + 10, "작성기관: Global Path AI – Market Intelligence Unit")
    c.drawRightString(w - 40, base_y, "저작권: © Global Path AI. All Rights Reserved.")

    c.save()
```
